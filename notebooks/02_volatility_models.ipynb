# 02 — Volatility Models

**Goal:** fit and compare four volatility forecasting approaches on the portfolio return series.

**Plan:**
1. Historical (rolling-window) volatility — the naive benchmark
2. EWMA (RiskMetrics-style)
3. GARCH(1,1)
4. EGARCH (captures the leverage/asymmetry effect)
5. Split into train/test, forecast out-of-sample
6. Compare RMSE, MAE and QLIKE loss across all four models

**Research question:** does GARCH-family modelling actually outperform simpler benchmarks out-of-sample, or does the added complexity not pay off on this portfolio?

```python
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from src.volatility import (
    historical_volatility, ewma_volatility, fit_garch, fit_egarch,
    forecast_volatility, evaluate_forecasts
)

port_returns = pd.read_csv('../data/portfolio_returns.csv', index_col=0, parse_dates=True).iloc[:, 0]
port_returns.tail()
```
